## Experiment
Inspect the intermediate outputs and check the assumptions before interpreting them.

```python
from socialhar.synthetic import generate
from socialhar.schema import validate
from socialhar.temporal import aggregate_activity, posting_entropy
frame = validate(generate())
summary = aggregate_activity(frame, min_subjects=3)
summary.head()
```

```python
import matplotlib.pyplot as plt
counts = frame.timestamp.dt.hour.value_counts().sort_index()
counts.plot.bar(title='Synthetic posting hours (UTC)')
plt.xlabel('UTC posting hour')
plt.ylabel('Records')
plt.tight_layout()
plt.show()
print('Posting entropy:', posting_entropy(counts))
```

## Interpretation and limitations
Posting time is not the time an activity occurred. Timezones, scheduled posts and platform use can change apparent rhythms. Small-cell suppression is not a differential privacy guarantee.
