```python
frame.groupby('permission').size()
assert frame.record_id.is_unique
assert frame.subject_id.nunique() == 70
```

## Interpretation and limitations
Each label is balanced by construction. This is a software fixture, not a representative sample. Profile records share the same toy distribution as posts; real profiles require separate annotation guidelines.
