## Experiment
Inspect the intermediate outputs and check the assumptions before interpreting them.

```python
from socialhar.synthetic import generate
from socialhar.split import grouped_split
from socialhar.topics import fit_topics
train, test = grouped_split(generate())
vectorizer, topic_model, weights, topics = fit_topics(train.text)
for i, terms in enumerate(topics):
    print(i, ', '.join(terms))
```

```python
test_weights = topic_model.transform(vectorizer.transform(test.text))
print(test_weights.shape)
assert (test_weights >= 0).all()
```

## Interpretation and limitations
NMF topics are nonnegative word factors, not diagnoses or stable human traits. Review examples, repeat seeds and inspect topic stability before assigning topic names.
