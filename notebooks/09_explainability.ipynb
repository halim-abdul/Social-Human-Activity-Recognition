## Experiment
Inspect the intermediate outputs and check the assumptions before interpreting them.

```python
from socialhar.synthetic import generate
from socialhar.split import grouped_split
from socialhar.models import baseline, explain_linear
train, test = grouped_split(generate())
model = baseline().fit(train.text, train.label)
explanation = explain_linear(model, 'I am cooking vegetable soup for dinner')
explanation
```

```python
import numpy as np
assert np.isclose(explanation['logit'], explanation['intercept'] + explanation['contribution_sum'])
import pandas as pd
pd.DataFrame(explanation['features'])
```

## Interpretation and limitations
The signed feature values reconstruct a class logit. Correlated bigrams and unigrams can share contributions; they explain this fitted model and are not causal explanations of human behavior.
