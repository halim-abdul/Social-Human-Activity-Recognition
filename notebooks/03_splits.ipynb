## Experiment
Inspect the intermediate outputs and check the assumptions before interpreting them.

```python
from socialhar.synthetic import generate
from socialhar.schema import validate
from socialhar.privacy import prepare
from socialhar.split import grouped_split
frame = prepare(validate(generate()))
train, test = grouped_split(frame)
assert not set(train.subject_id) & set(test.subject_id)
print(len(train), len(test))
```

```python
from socialhar.split import chronological_split
past, future = chronological_split(frame, '2026-02-01T00:00:00Z')
print(past.timestamp.max(), future.timestamp.min())
```

## Interpretation and limitations
The grouped split asks whether the model generalizes to unseen authors. The chronological split asks about future records and can contain the same authors. Do not report them as equivalent estimates.
